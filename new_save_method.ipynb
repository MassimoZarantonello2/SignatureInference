```python
import json

def save_results(predictor, test_data, output_path='results_summary.json'):
    evaluations = predictor.evaluate(test_data)
    results = []

    for signature in evaluations:
        model_predictor = predictor.get_predictor(signature)
        metrics = evaluations[signature]
        info = model_predictor.info()
        model_info = info['model_info']

        best_model_name = info['best_model']
        best_model_info = model_info[best_model_name]

        ensemble_model_names = best_model_info.get('features', [])

        ensemble_info = [
            {
                'model_name': model,
                'val_score': model_info[model].get('val_score'),
                'fit_time': model_info[model].get('fit_time')
            }
            for model in ensemble_model_names
        ]

        result = {
            'signature': signature,
            'best_model': best_model_name,
            'best_model_val_score': best_model_info.get('val_score'),
            'best_model_metrics': metrics,
            'best_model_fit_time': best_model_info.get('fit_time'),
            'ensemble_models': ensemble_model_names,
            'ensemble_info': ensemble_info
        }

        results.append(result)

    # Salvataggio su file JSON
    with open(output_path, 'w') as f:
        json.dump(results, f, indent=4)  # indent=4 per renderlo leggibile

    print(f"✅ Risultati salvati in '{output_path}'")

    return results
```

```python
results_summary = save_results(ml_predictor, test_data, output_path='results_summary.json')
```

### Final Results

```python
ml_results_df = pd.DataFrame(columns=['signature', 'best_model', 'accuracy'])
for i in range(len(gt_labels)):
    predictor_class = ml_predictor.get_predictor(gt_labels[i])
    ml_results_df.loc[i, 'signature'] = gt_labels[i]
    ml_results_df.loc[i, 'best_model'] = predictor_class.leaderboard().iloc[0]['model']
    ml_results_df.loc[i, 'accuracy'] = predictor_class.leaderboard().iloc[0]['score_val']
    
    
ml_results_df.head(29)
```

```python
results_df = pd.DataFrame(columns=['signature', 'best_model', 'accuracy'])
for i in range(len(gt_labels)):
    predictor_class = predictor.get_predictor(gt_labels[i])
    results_df.loc[i, 'signature'] = gt_labels[i]
    results_df.loc[i, 'best_model'] = predictor_class.leaderboard().iloc[0]['model']
    results_df.loc[i, 'accuracy'] = predictor_class.leaderboard().iloc[0]['score_val']
    
    
results_df.head(29)
```

```python
final_df = pd.DataFrame(columns=['signature', 'best_model', 'accuracy'])
for i in range(len(gt_labels)):
    final_df.loc[i, 'signature'] = gt_labels[i]
    if ml_results_df.iloc[i]['best_model'] == results_df.iloc[i]['best_model']:
        final_df.loc[i, 'best_model'] = ml_results_df.iloc[i]['best_model']
    else:
        final_df.loc[i, 'best_model'] = ml_results_df.iloc[i]['best_model'] + ' vs ' + results_df.iloc[i]['best_model']
    final_df.loc[i, 'accuracy'] = ml_results_df.iloc[i]['accuracy'] - results_df.iloc[i]['accuracy']
```
